```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x_axis = []
y_axis = []
idx = 0
with open('../../../NOP/NOP_params/visualization/exec.txt', 'r') as file:
    for line in file:
        y_axis.append(float(line.strip()))
        x_axis.append(idx)
        idx+=1

plt.rc('font', size=8)
plt.yticks(np.arange(0.05, max(y_axis)+0.05, step=0.005))
plt.xlabel('No. of Iterations')
plt.ylabel('Time Elapsed (milliseconds)')
plt.text(36, 0.128, 'Representation of NOP execution with parameters')
plt.plot(x_axis,y_axis)
plt.show()
```

```python
x_axis_2 = []
y_axis_2 = []
idx_2 = 0
with open('../../../NOP/NOP_params/visualization/exec.txt', 'r') as file_2:
    for line in file_2:
        y_axis_2.append(float(line.strip()))
        x_axis_2.append(idx_2)
        idx_2+=1
plt.rc('font', size=8)
plt.yticks(np.arange(0.05, max(y_axis_2)+0.05, step=0.005))
plt.xlabel('No. of Iterations')
plt.ylabel('Time Elapsed (milliseconds)')
plt.text(36, 0.128, 'Representation of NOP execution with no parameters')
plt.plot(x_axis_2,y_axis_2)
plt.show()
```
